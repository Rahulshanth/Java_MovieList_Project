```python
# Define the IQR-based outlier detection function from Code Reuse Session 1
def find_outliers_IQR(data):
    # Calculate the 25th percentile (Q1) of the data
    q1 = data.quantile(0.25)
    # Calculate the 75th percentile (Q3) of the data
    q3 = data.quantile(0.75)
    # Calculate the Inter-Quartile Range (IQR = Q3 - Q1)
    IQR = q3 - q1
    # Identify outliers: values below Q1-1.5*IQR or above Q3+1.5*IQR
    outliers = data[((data < (q1 - 1.5 * IQR)) | (data > (q3 + 1.5 * IQR)))]
    # Return the outlier values
    return outliers
```

```python
# ── Apply IQR outlier detection to the 'age' column ──
# Detect outliers in the age column using the IQR function
age_outliers = find_outliers_IQR(data['age'])
# Print the total number of IQR-detected outliers in age
print("Number of IQR outliers in 'age':", len(age_outliers))
# Display the actual outlier values for inspection
print(age_outliers)
print()
# NOTE: IQR flags many valid ages (>40) because the age distribution is young-skewed.
# Only age=123 is genuinely erroneous and will be removed using a threshold filter.
```

```python
# ── Apply IQR outlier detection to the 'loan_interest_rate' column ──
# Detect outliers in loan_interest_rate using the IQR function
rate_outliers = find_outliers_IQR(data['loan_interest_rate'])
# Print the number of IQR outliers detected in loan interest rate
print("Number of IQR outliers in 'loan_interest_rate':", len(rate_outliers))
# Display the actual outlier values
print(rate_outliers)
print()
# NOTE: A negative interest rate (-11.14) is clearly erroneous and will be filtered out.
```

```python
# ── Apply IQR outlier detection to the 'emplyment_length' column ──
# Detect outliers in employment length using the IQR function
emp_outliers = find_outliers_IQR(data['emplyment_length'])
# Print the number of IQR outliers detected in employment length
print("Number of IQR outliers in 'emplyment_length':", len(emp_outliers))
# Display the top and bottom of the outlier list
print(emp_outliers.sort_values().tail(10))
print()
# NOTE: Values of 123 and 150 years of employment are clearly impossible and will be removed.
```

```python
# ── Apply IQR outlier detection to 'income', 'loan_amount', 'loan_income_ratio', 'credit_history_length' ──
# Detect outliers in income column
income_outliers = find_outliers_IQR(data['income'])
# Print the count of income IQR outliers (high earners - kept as legitimate values)
print(f"IQR outliers in 'income': {len(income_outliers)} — These are likely legitimate high-income clients. RETAINED.")

# Detect outliers in loan_amount column
loan_outliers = find_outliers_IQR(data['loan_amount'])
# Print the count of loan_amount IQR outliers
print(f"IQR outliers in 'loan_amount': {len(loan_outliers)} — All values within realistic loan range. RETAINED.")

# Detect outliers in loan_income_ratio column
ratio_outliers = find_outliers_IQR(data['loan_income_ratio'])
# Print the count of loan_income_ratio IQR outliers
print(f"IQR outliers in 'loan_income_ratio': {len(ratio_outliers)} — Values between 0 and 0.83, all realistic. RETAINED.")

# Detect outliers in credit_history_length column
credit_outliers = find_outliers_IQR(data['credit_history_length'])
# Print the count of credit_history_length IQR outliers
print(f"IQR outliers in 'credit_history_length': {len(credit_outliers)} — Values up to 30 years, all realistic. RETAINED.")
```

```python
# Create a copy of the original dataset to avoid modifying the raw data
data_clf = data.copy()

# Drop the 'id' column as it is just a unique identifier with no predictive value
data_clf = data_clf.drop(columns=(['id', 'max_allowed_loan']))

# Confirm the remaining columns after dropping irrelevant ones
print("Columns after dropping 'id' and 'max_allowed_loan':")
print(list(data_clf.columns))
print(f"Shape: {data_clf.shape}")
```

```python
# Print shape before removing erroneous outliers to show before state
print(f"Shape BEFORE removing erroneous outliers: {data_clf.shape}")

# Remove the erroneous age outlier: age=123 is biologically impossible
# Filter to keep only rows where age is less than 100
data_clf = data_clf[data_clf.age < 100]

# Remove negative loan interest rate: interest rates cannot be negative
# Filter to keep only rows where loan_interest_rate is greater than 0
data_clf = data_clf[data_clf.loan_interest_rate > 0]

# Remove impossible employment lengths: 123 and 150 years are not realistic
# Filter to keep only rows where employment length is less than 50 years
data_clf = data_clf[data_clf.emplyment_length < 50]

# Print shape after removing erroneous outliers to show after state
print(f"Shape AFTER removing erroneous outliers: {data_clf.shape}")

# Verify issue 2: confirm erroneous age outlier has been removed
#print(f"Max age after cleaning: {data_clf['age'].max()}")
#print(f"Rows with age > 100: {(data_clf['age'] > 100).sum()}")

# Verify Issue 3: negative interest rate removed
#print(f"Min loan_interest_rate after cleaning: {data_clf['loan_interest_rate'].min():.2f}")
#print(f"Rows with rate < 0: {(data_clf['loan_interest_rate'] < 0).sum()}")

# Verify Issue 4: impossible employment length removed
#print(f"Max emplyment_length after cleaning: {data_clf['emplyment_length'].max()}")
#print(f"Rows with emplyment_length > 50: {(data_clf['emplyment_length'] > 50).sum()}")

```

```python
# ── Before: check missing values ──
print("Missing values BEFORE imputation:")
print(data_clf.isnull().sum())
print()
```

```python
# Calculate the mean of the 'age' column (excluding NaN values by default)
Mean_age = data_clf['age'].mean()
# Fill missing values in 'age' with the calculated mean
data_clf['age'].fillna(Mean_age, inplace=True)
print(f"'age' missing values filled with mean: {Mean_age:.2f}")
```

```python
# Calculate the mean of the 'loan_interest_rate' column
Mean_loan_interest_rate = data_clf['loan_interest_rate'].mean()
# Fill missing values in 'loan_interest_rate' with the calculated mean
data_clf['loan_interest_rate'].fillna(Mean_loan_interest_rate, inplace=True)
print(f"'loan_interest_rate' missing values filled with mean: {Mean_loan_interest_rate:.2f}")
```

```python
# Drop rows where 'payment_default_on_file' is missing (only 5 rows, <0.01% of data)
# This avoids bias from imputing a binary categorical variable
data_clf = data_clf.dropna(subset=['payment_default_on_file'])

# Confirm no missing values remain after cleaning
print("Missing values AFTER all cleaning steps:")
print(data_clf.isnull().sum())
print(f"Shape after dropping missing payment_default_on_file rows: {data_clf.shape}")
```

```python
# Print unique values before encoding to show before state
print("payment_default_on_file BEFORE encoding:", data_clf['payment_default_on_file'].unique())

# Map 'Y' (Yes, has defaulted) to 1 and 'N' (No default) to 0 using binary mapping
data_clf['payment_default_on_file'] = data_clf['payment_default_on_file'].map({'Y': 1, 'N': 0})

# Print unique values after encoding to confirm transformation
print("payment_default_on_file AFTER encoding:", data_clf['payment_default_on_file'].unique())
```

```python
# Initialise the LabelEncoder from sklearn preprocessing
label_encoder = LabelEncoder()

# Print unique values of home_ownership before encoding
print("home_ownership BEFORE encoding:", data_clf['home_ownership'].unique())
# Apply LabelEncoder to 'home_ownership': converts string categories to integer labels
data_clf['home_ownership'] = label_encoder.fit_transform(data_clf['home_ownership'])
# Print unique values after encoding to confirm transformation
print("home_ownership AFTER encoding:", data_clf['home_ownership'].unique())
# Print the mapping of labels for reference
print("home_ownership mapping:", dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_))))
print()

# Print unique values of loan_intent before encoding
print("loan_intent BEFORE encoding:", data_clf['loan_intent'].unique())
# Apply LabelEncoder to 'loan_intent': converts string categories to integer labels
data_clf['loan_intent'] = label_encoder.fit_transform(data_clf['loan_intent'])
# Print unique values after encoding to confirm transformation
print("loan_intent AFTER encoding:", data_clf['loan_intent'].unique())
# Print the mapping of labels for reference
print("loan_intent mapping:", dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_))))
```

```python
# Convert float columns to integer where appropriate (age and loan_interest_rate were float due to NaN)
data_clf = data_clf.astype({
    # Convert age to integer now that NaN values have been filled
    "age": "int",
    # Convert loan_interest_rate to float (keep as float - it's a percentage with decimal places)
    "loan_interest_rate": "float",
    # Convert payment_default_on_file to integer (was mapped to 0 or 1)
    "payment_default_on_file": "int"
})

# Verify the final data types of all columns
print("Final data types for Classification Dataset:")
print(data_clf.dtypes)
```

```python
# Define the list of numeric input columns (continuous variables to be scaled)
numeric_cols_clf = ['age', 'income', 'emplyment_length', 'loan_amount',
                    'loan_interest_rate', 'loan_income_ratio', 'credit_history_length']

# Define the list of encoded categorical columns (already converted to integers)
categorical_cols_clf = ['home_ownership', 'loan_intent', 'payment_default_on_file']

# Print both lists for confirmation
print("Numeric columns (to be scaled):", numeric_cols_clf)
print("Categorical columns (encoded):", categorical_cols_clf)
# The target column 'loan_approval_status' is NOT included in either list (it is the output y)
print("Target column: 'loan_approval_status'")
```

```python
# Print descriptive stats BEFORE scaling to show the original scale
print("Numeric features BEFORE MinMax Scaling:")
print(data_clf[numeric_cols_clf].describe().round(4))
print()

# Initialise the MinMaxScaler to scale values to the range [0, 1]
scaler = MinMaxScaler()

# Apply MinMaxScaler to all numeric input columns only (NOT to the target variable)
data_clf[numeric_cols_clf] = scaler.fit_transform(data_clf[numeric_cols_clf])

# Print descriptive stats AFTER scaling to confirm values are now in [0, 1] range
print("Numeric features AFTER MinMax Scaling (all values should be between 0 and 1):")
print(data_clf[numeric_cols_clf].describe().round(4))
```

```python
# Display the final shape of the classification dataset
print(f"Final Classification Dataset Shape: {data_clf.shape}")
print()

# Display the final column list
print("Final columns:", list(data_clf.columns))
print()

# Confirm no missing values remain in the final dataset
print("Missing values in final classification dataset:")
print(data_clf.isnull().sum())
print()

# Preview the first 5 rows of the cleaned, prepared classification dataset
print("Preview of final classification dataset:")
data_clf.head()
```
